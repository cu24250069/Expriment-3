```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
image_path = list(uploaded.keys())[0]

image = cv2.imread(image_path)

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
gaussian = cv2.GaussianBlur(image, (5, 5), 0)

gaussian_rgb = cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(gaussian_rgb)
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```python
median = cv2.medianBlur(image, 5)

median_rgb = cv2.cvtColor(median, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(median_rgb)
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```python
average = cv2.blur(image, (5, 5))

average_rgb = cv2.cvtColor(average, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(average_rgb)
plt.title("Average Filter")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

laplacian = cv2.Laplacian(gray, cv2.CV_64F)

plt.figure(figsize=(6, 6))
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian Edge Detection")
plt.axis("off")
plt.show()
```

```python
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
```

```python
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X - Vertical Edges")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sobel_y, cmap="gray")
plt.title("Sobel Y - Horizontal Edges")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(gaussian_rgb)
plt.title("Gaussian Filter")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(median_rgb)
plt.title("Median Filter")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(average_rgb)
plt.title("Average Filter")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(sobel_x, cmap="gray")
plt.title("Sobel X")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Conclusion

The experiment successfully demonstrated the implementation of different spatial filtering techniques using Python and OpenCV. Low-pass filters such as Average, Gaussian, and Median filters were used for image smoothing and noise reduction. Gaussian filtering provided smooth results while preserving important image structures, whereas Median filtering was particularly effective for removing impulse noise. High-pass filters such as Laplacian and Sobel operators successfully highlighted edges and fine details. The experiment showed that spatial filtering is an important preprocessing technique in computer vision and image analysis applications.
